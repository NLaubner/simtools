```python
import sys
from pathlib import Path

sys.path.append(str(Path.cwd().parent))

from src.simulation import load_config, run_monte_carlo
```

```python
cfg = load_config("../configs/params.json")
results, summary = run_monte_carlo(cfg, return_details=True)

summary
```

```python
import matplotlib.pyplot as plt

spot_data = summary[summary["metric"].str.startswith("p_infection_happens_at_")]

spots = spot_data["metric"].str.replace("p_infection_happens_at_", "")
values = spot_data["value"]

plt.figure(figsize=(6,4))
plt.bar(spots, values)
plt.xlabel("location")
plt.ylabel("Infection Probability")
plt.title("Probability of Infection sorted by location")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

```
